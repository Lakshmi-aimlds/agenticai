```python
from langchain_community.tools import WikipediaQueryRun
```

```python
from langchain_community.utilities import WikipediaAPIWrapper
```

```python
api_wrapper = WikipediaAPIWrapper(top_k_results=5, doc_content_char_max=500)
```

```python
wiki_tool = WikipediaQueryRun(api_wrapper=api_wrapper, name="Wikipedia Query Run", description="A tool to query Wikipedia for information.")
```

```python
wiki_tool.name
```

```python
wiki_tool.args
```

```python
wiki_tool.run({"query":"What is LangGraph?"})  #Example query
```

### CUSTOM TOOLS

```python
from langchain.agents import tool

@tool
def multiply(a:int, b:int) -> int:
    """This tool is for multiplication"""
    return a*b    
```

```python
multiply.invoke({"a":10, "b":5})
```
